```python
from torchvision.models import vit_b_16
```

```python
def compute_parameters(layer):
    return sum(p.numel() for p in layer.parameters())
```

```python
model = vit_b_16(weights=None)
```

```python
model
```

```python
for name, layer in model.named_children():
    print(f"{name:20} : {compute_parameters(layer) * 4 / (1024 ** 2):.2f} MB")

print(f"{'Total':20} : {compute_parameters(model):10,} Parameters")
print(f"Size of Model: {compute_parameters(model) * 4 / (1024 ** 2):.2f} MB")
```

```python
for name, layer in model.encoder.layers.named_children():
    print(f"{name:20} : {compute_parameters(layer) * 4 / (1024 ** 2):.2f} MB")
    for sub_name, sub_layer in layer.named_children():
        print(f" > {sub_name:18} : {compute_parameters(sub_layer) * 4 / (1024 ** 2):.2f} MB")
    print("-"*50)
```
